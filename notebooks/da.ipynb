```python
import pandas as pd

df = pd.read_excel("../data/online-retail-dataset.xlsx")
```

```python
df.info()
```

```python
null_info = df.isnull().sum().reset_index()
null_info.columns = ['Column', 'Null Count']
null_info = null_info[null_info['Null Count'] > 0]
null_info
```

```python
duplicates = df[df.duplicated()]
duplicates
```

```python
cdf = df.drop_duplicates()
cdf
```

```python
cdf[cdf.duplicated()]
```

```python
null_info = cdf.isnull().sum().reset_index()
null_info.columns = ['Column', 'Null Count']
null_info = null_info[null_info['Null Count'] > 0]
null_info
```

```python
null_rows = cdf[cdf.isnull().any(axis=1)]
null_rows
```

```python
df_returns = cdf[cdf['Quantity'] < 0]
df_returns
```

```python
cdf_credits = cdf[cdf['UnitPrice'] < 0]
cdf_credits
```

```python
# Add Sales column
cdf['Sales'] = cdf['Quantity'] * cdf['UnitPrice']
cdf.sort_values('Sales', ascending=False).head(10)
```

```python
top_selling_by_revenue = cdf.groupby('Description')['Sales'].sum().sort_values(ascending=False).head(10)
top_selling_by_revenue
```

```python
df_purchases = cdf[cdf['Quantity'] > 0].copy()
top_selling_items = df_purchases.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)
top_selling_items
```

```python
df_returns = cdf[cdf['Quantity'] < 0].copy()
most_returned_items = df_returns.groupby('Description')['Quantity'].sum().abs().sort_values(ascending=False).head(10)
most_returned_items
```

```python
top_purchasing_customer = cdf.groupby('CustomerID')['Sales'].sum().abs().sort_values(ascending=False)
top_purchasing_customer
```
